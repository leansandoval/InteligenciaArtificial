Las imágenes anteriores pertenecen a las soluciones visuales simuladas para algunos ejercicios con algoritmos genéticos:

* Convergencia del Fitness: muestra cómo mejora la aptitud para el problema de uso de máquinas en fábrica y consumo energético urbano.
* Varianza Genética por Gen: observa la diversidad a lo largo de generaciones para los genes de asignación de máquinas.
* Frontera de Pareto: para el problema multiobjetivo de eficiencia energética vs emisiones en una ciudad.
* Matriz de Calor: representa cómo se asignan las máquinas a tareas (corte, ensamble, pintura) en la fábrica.
