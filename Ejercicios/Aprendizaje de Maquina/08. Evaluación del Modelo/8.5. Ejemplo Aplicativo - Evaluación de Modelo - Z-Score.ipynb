```python
import numpy as np
import pandas as pd
from scipy.stats import zscore
```

```python
# Generar datos de ejemplo con outliers
data = np.array([10, 12, 14, 15, 13, 12, 300, 14, 11, 15, 13, 1000])  # Hay dos outliers (300 y 1000)
```

```python
# Convertir a DataFrame
df = pd.DataFrame(data, columns=['Valor'])
```

```python
# Calcular Z-Score
df['Z-Score'] = zscore(df['Valor'])
```

```python
# Identificar valores atípicos (|Z| > 3)
df['Es_Outlier'] = df['Z-Score'].abs() > 3
```

```python
print(df)
```

Explicación del código:

* Se crea un conjunto de datos con valores normales y algunos outliers.
* Se calcula el Z-Score de cada dato.
* Se marca como outlier cualquier dato con |Z| > 3.
