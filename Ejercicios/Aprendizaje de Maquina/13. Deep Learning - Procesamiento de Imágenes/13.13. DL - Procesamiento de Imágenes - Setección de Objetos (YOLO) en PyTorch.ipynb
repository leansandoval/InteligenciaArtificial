En este ejercicio se usa YOLO (You Only Look Once), un modelo eficiente para la detección de objetos. En PyTorch, puedes cargar un modelo preentrenado de YOLO para realizar detección de objetos en imágenes.

```python
# Importación de librerías
import torch
import cv2
from matplotlib import pyplot as plt
```

```python
# Cargar el modelo preentrenado de YOLOv5
model = torch.hub.load('ultralytics/yolov5', 'yolov5s')  # 'yolov5s' es el modelo más pequeño
```

```python
# Cargar una imagen
img = cv2.imread('path/to/image.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
# Realizar detección de objetos
results = model(img)
```

```python
# Mostrar los resultados
results.show()  # Muestra la imagen con las predicciones de las cajas delimitadoras
```

```python
# Obtener las coordenadas de las cajas delimitadoras y las clases
boxes = results.xywh[0]  # Coordenadas de las cajas [x, y, w, h, clase]
labels = results.names  # Nombres de las clases detectadas
```
