Este ejercicio usa un modelo de red neuronal convolucional (CNN) básico para clasificar imágenes en categorías. Puedes entrenar el modelo utilizando un conjunto de datos de imágenes de tu elección.

```python
# Importación de librerías
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
```

```python
# Definir transformaciones de los datos
transform = transforms.Compose([
    transforms.Resize((150, 150)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
```

```python
# Cargar el conjunto de datos
train_data = datasets.ImageFolder('path/to/train/data', transform=transform)
train_loader = DataLoader(train_data, batch_size=32, shuffle=True)

val_data = datasets.ImageFolder('path/to/val/data', transform=transform)
val_loader = DataLoader(val_data, batch_size=32, shuffle=False)
```

```python
# Usar un modelo preentrenado (ResNet)
model = models.resnet50(pretrained=True)
```

```python
# Congelar las capas preentrenadas
for param in model.parameters():
    param.requires_grad = False
```

```python
# Modificar la última capa para clasificación binaria
model.fc = nn.Linear(model.fc.in_features, 2)
```

```python
# Definir optimizador y función de pérdida
optimizer = optim.Adam(model.fc.parameters(), lr=0.001)
criterion = nn.CrossEntropyLoss()
```

```python
# Entrenamiento del modelo
epochs = 10
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for images, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
    print(f"Epoch {epoch+1}/{epochs}, Loss: {running_loss/len(train_loader)}")
```

```python
# Validación del modelo
model.eval()
correct = 0
total = 0
with torch.no_grad():
    for images, labels in val_loader:
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()
```

```python
print(f"Accuracy: {100 * correct / total}%")
```
