Aquí te dejo el código para trabajar con imágenes satelitales en formato GeoTIFF utilizando rasterio para leer los datos y luego procesarlos con Keras. Este código es útil si tienes imágenes geoespaciales y deseas clasificarlas con una red neuronal convolucional.

```python
# Importación de librerías
import tensorflow as tf
import rasterio
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt

# Función para leer imágenes GeoTIFF
def load_geotiff(image_path):
    with rasterio.open(image_path) as src:
        img_array = src.read([1, 2, 3])  # Leer las bandas RGB
        img_array = np.moveaxis(img_array, 0, -1)  # Cambiar el orden para que sea (height, width, channels)
    return img_array

# Cargar una imagen GeoTIFF de ejemplo
img = load_geotiff('path/to/your/geotiff_image.tif')

# Normalizar la imagen
img = img / 255.0

# Visualizar la imagen
plt.imshow(img)
plt.show()

# Definir el modelo CNN para clasificación
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(img.shape[0], img.shape[1], 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.Dense(1, activation='sigmoid')  # Para clasificación binaria
])

# Compilar el modelo
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Suponiendo que tienes datos de entrenamiento (imágenes y etiquetas)
# Aquí te doy un ejemplo para entrenar
train_images = np.array([img])  # Lista de imágenes de entrenamiento
train_labels = np.array([1])   # Etiquetas correspondientes

# Entrenar el modelo
model.fit(train_images, train_labels, epochs=5)
```
