Podemos observar tres ejemplos visuales que comparan distintos comportamientos de modelos de regresión respecto a sesgo y varianza:

1. Alto Sesgo (Modelo Lineal)
* MSE Train/Test altos
* La línea roja no logra capturar la forma real (sinusoide) de los datos.
Diagnóstico: Underfitting → el modelo es demasiado simple → alto sesgo

2. Alta Varianza (Polinomio Grado 15)
* MSE Train muy bajo, pero MSE Test muy alto
* Se ajusta perfectamente al ruido del entrenamiento, pero falla en generalizar.
Diagnóstico: Overfitting → el modelo es muy complejo → alta varianza

3. Balanceado (Polinomio Grado 4)
* MSE Train y Test razonables y similares
* Captura bien la tendencia sin sobreajustar el ruido.
Diagnóstico: Buen equilibrio entre sesgo y varianza.
