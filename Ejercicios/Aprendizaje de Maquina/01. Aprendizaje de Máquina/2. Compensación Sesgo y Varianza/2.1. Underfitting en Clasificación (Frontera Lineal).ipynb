```python
# Ejemplo para clasificación (Underfitting)
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.linear_model import LogisticRegression
```

```python
X, y = make_moons(n_samples=100, noise=0.2, random_state=42)
```

```python
model = LogisticRegression()
model.fit(X, y)
```

```python
# Frontera de decisión
xx, yy = np.mgrid[-1.5:2.5:.01, -1:1.5:.01]
grid = np.c_[xx.ravel(), yy.ravel()]
probs = model.predict_proba(grid)[:, 1].reshape(xx.shape)

plt.contourf(xx, yy, probs, levels=[0, 0.5, 1], cmap='RdBu', alpha=0.6)
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='RdBu', edgecolors='white')
plt.title("Underfitting en Clasificación (Frontera Lineal)")
plt.show()
```
