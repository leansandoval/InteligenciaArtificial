```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Ejemplo: Curva de error para overfitting
complexity = np.arange(1, 10)
train_error = np.array([0.9, 0.7, 0.5, 0.3, 0.1, 0.05, 0.03, 0.02, 0.01])
val_error = np.array([0.9, 0.8, 0.7, 0.6, 0.5, 0.6, 0.7, 0.8, 0.9])
```

```python
plt.plot(complexity, train_error, label="Entrenamiento")
plt.plot(complexity, val_error, label="Evaluación")
plt.xlabel("Complejidad (Grado del polinomio)")
plt.ylabel("Error")
plt.legend()
plt.title("Diagnóstico: Overfitting")
plt.show()
```
