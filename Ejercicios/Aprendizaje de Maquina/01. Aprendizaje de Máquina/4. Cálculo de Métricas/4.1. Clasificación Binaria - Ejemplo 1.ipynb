```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
```

```python
# Etiquetas verdaderas y predichas
y_true = [1]*50 + [0]*50
y_pred = [1]*40 + [0]*10 + [1]*5 + [0]*45
```

image.png

```python
# Métricas
print("Accuracy:", accuracy_score(y_true, y_pred))
print("Precision:", precision_score(y_true, y_pred))
print("Recall:", recall_score(y_true, y_pred))
print("F1-Score:", f1_score(y_true, y_pred))
```

Interpretación - Respecto a Clase 1 (Spam):
* Accuracy = (TP + TN) / Total = 85%
* Precision = 40 / (40 + 5) = 0.888
* Recall = 40 / (40 + 10) = 0.80
* F1-score ≈ 0.842

Análisis de Error:
* El modelo comete más errores al clasificar spam como no spam (10 FNs), lo que sugiere que recall es más bajo que la precisión.

Sesgo/Varianza:
* Si en entrenamiento el modelo acierta todo (accuracy ≈ 1.0) pero en evaluación baja (≈ 0.85), podríamos sospechar sobreajuste (alta varianza).
* Si en ambos es baja (por ejemplo, 0.60), se trataría de alto sesgo.
