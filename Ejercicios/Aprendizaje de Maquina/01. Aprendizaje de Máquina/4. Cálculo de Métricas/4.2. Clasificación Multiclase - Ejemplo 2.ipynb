```python
from sklearn.metrics import classification_report
```

```python
y_true = [0]*50 + [1]*50 + [2]*50
y_pred = (
    [0]*45 + [1]*5 +        # para clase 0
    [0]*4 + [1]*40 + [2]*6 +# para clase 1
    [0]*1 + [1]*3 + [2]*46  # para clase 2
)
```

image.png

```python
print(classification_report(y_true, y_pred, target_names=['remera', 'pantalón', 'zapato']))
```

Interpretación:
* Clase 1 (pantalón) tiene el peor desempeño (6 predicciones mal clasificadas como zapatos).
* Clase 0 (remera) está muy bien predicha.
* El modelo puede tener confusión entre clases similares (pantalón y zapato).

Análisis de Error:
* El mayor error ocurre en las clases 1 y 2.
* Verificar si estas clases tienen características visuales similares.

Sesgo/Varianza:
* Si estas confusiones ocurren en entrenamiento y evaluación, puede ser alto sesgo (modelo poco expresivo).
* Si ocurre solo en evaluación, es señal de alta varianza (sobreajuste).
