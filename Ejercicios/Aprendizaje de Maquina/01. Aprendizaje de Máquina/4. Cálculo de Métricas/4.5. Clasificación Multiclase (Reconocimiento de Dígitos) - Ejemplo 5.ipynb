🔹 Resultados (con classification_report):

```python
from sklearn.metrics import classification_report
```

```python
y_true = [0]*20 + [1]*20 + [2]*20 + [3]*20 + [4]*20
y_pred = (
    [0]*18 + [1]*1 + [2]*1 +            # Real 0
    [0]*2 + [1]*15 + [2]*1 + [3]*2 +    # Real 1
    [2]*17 + [3]*3 +                    # Real 2
    [1]*1 + [2]*2 + [3]*14 + [4]*3 +    # Real 3
    [3]*1 + [4]*19                      # Real 4
)
```

```python
print(classification_report(y_true, y_pred, digits=3))
```

Análisis:
* Clase 3 es la más confusa → errores con clases 2 y 4.
* Clase 0 y 4 tienen mejor desempeño.

Diagnóstico:
* Si tanto en entrenamiento como en evaluación se mantiene esta confusión → alto sesgo.
* Si en entrenamiento es perfecto y esto ocurre en evaluación → alta varianza.
