La fórmula general para el tamaño de salida es:

Salida=⌊(𝑊−𝐾+2𝑃)/𝑆⌋+1

donde:
* 𝑊: dimensión de entrada (asumimos cuadrada)
* 𝐾: tamaño del kernel
* 𝑆: stride
* 𝑃: padding

Ejercicio – Cálculo general por capas

Enunciado:

Arquitectura CNN parcial:
* Input: 64×64×3
* Conv1: 32 filtros, kernel 3×3, stride 1, padding 1
* Conv2: 64 filtros, kernel 3×3, stride 2, padding 1

Solución:

Después de Conv1:
* Salida=[(64−3+2*(1))/1]+1=64
* Tamaño: 64×64×32

Después de Conv2:
* Salida=[(64−3+2*(1))/2]+1=(63/2)+1=31.5+1≈32
* Tamaño: 32×32×64
