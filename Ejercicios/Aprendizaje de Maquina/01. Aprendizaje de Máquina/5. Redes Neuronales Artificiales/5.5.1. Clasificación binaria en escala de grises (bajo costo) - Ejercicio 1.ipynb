Enunciado:

Estás diseñando un sistema embebido para distinguir si una cámara de seguridad detecta persona o no.
Las imágenes son en escala de grises de 96×96 px.
Se necesita eficiencia en tiempo real y bajo consumo.

Solución propuesta:
* Tipo de convolución: Conv2D, ya que la entrada tiene 1 canal (gris)
* Capas de agrupamiento: MaxPooling2D después de cada bloque conv
* Funciones de activación: ReLU para capas ocultas, Sigmoid para salida
* Capas ocultas: 2 bloques convolucionales, 1 densa final
* Profundidad: baja (por eficiencia)

Arquitectura sugerida:
* Input: 96×96×1
* Conv1: 3×3, pad=1, stride=1, 16 filtros → 96×96×16  
  * MaxPool1: 2×2 → 48×48×16  
* Conv2: 3×3, pad=1, stride=1, 32 filtros → 48×48×32  
  * MaxPool2: 2×2 → 24×24×32  
* Flatten → Dense(64) → Dense(1, activation='sigmoid')
