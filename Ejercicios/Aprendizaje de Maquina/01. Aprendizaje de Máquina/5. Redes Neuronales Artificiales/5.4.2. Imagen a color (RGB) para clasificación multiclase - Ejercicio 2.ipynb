Enunciado:

Queremos clasificar imágenes a color
128×128×3 en 10 clases distintas. Se busca un modelo CNN más profundo para buena precisión.

Análisis y decisiones de diseño:
* Tipo de convolución:
  * Conv2D, entrada con 3 canales (RGB).
* Pooling:
  * MaxPooling para reducir resolución, cada 2 bloques convolucionales.
* Funciones de activación:
  * ReLU (conv, dense), Softmax (salida multiclase)
* Capas ocultas y profundidad:
  * Mínimo 3 bloques convolucionales, luego Dense.

Cálculo de dimensiones paso a paso (supongamos 32-64-128 filtros):
* Input: 128×128×3
* Conv1 (3×3, pad=1, stride=1, 32 filtros): → 128×128×32
  * MaxPool1 (2×2): → 64×64×32
* Conv2 (3×3, 64 filtros): → 64×64×64
  * MaxPool2: → 32×32×64
* Conv3 (3×3, 128 filtros): → 32×32×128
  * MaxPool3: → 16×16×128
* Flatten: → 16×16×128 = 32,768
* Dense(256) + Dropout + Dense(10, softmax)
