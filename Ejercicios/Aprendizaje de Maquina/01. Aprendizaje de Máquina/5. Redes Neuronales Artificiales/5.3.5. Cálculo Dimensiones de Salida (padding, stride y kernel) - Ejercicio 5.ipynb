La fórmula general para el tamaño de salida es:

Salida=⌊(𝑊−𝐾+2𝑃)/𝑆⌋+1

donde:
* 𝑊: dimensión de entrada (asumimos cuadrada)
* 𝐾: tamaño del kernel
* 𝑆: stride
* 𝑃: padding

Imagen color con stride > 1 y padding 0

Enunciado:
* Imagen: 128×128×3
* Filtro: 5×5×3
* Stride: 2
* Padding: 0
* N° de filtros: 32

Solución:

Salida=⌊(128−5+2×0)/2⌋+1=⌊123/2⌋+1=61+1=62

Salida: 62×62×32
