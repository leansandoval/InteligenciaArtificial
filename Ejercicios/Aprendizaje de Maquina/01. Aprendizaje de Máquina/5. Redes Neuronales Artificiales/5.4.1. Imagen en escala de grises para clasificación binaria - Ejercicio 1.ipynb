Enunciado:

Queremos clasificar imágenes en escala de grises de tamaño 64×64 px en 2 clases: “gato” o “perro”.

Se busca una red CNN simple y eficiente para dispositivos de bajo costo.

Análisis y decisiones de diseño:
* Tipo de convolución: Conv2D, ya que es imagen en escala de grises → entrada con 1 canal.
* Capas de agrupamiento: MaxPooling2D con stride 2 después de cada bloque conv → reduce dimensión y parámetros.
* Funciones de activación:
  * ReLU en capas convolucionales y ocultas.
  * Sigmoid al final para clasificación binaria.
* Capas ocultas y profundidad:
  * Poca profundidad por eficiencia: 2 bloques conv+pool y 1 capa densa.
* Cálculo de dimensiones (supongamos):
  * Conv1: kernel 3×3, padding=1, stride=1 → salida: 64×64
    * MaxPool1: kernel=2, stride=2 → salida: 32×32
  * Conv2: kernel 3×3, padding=1, stride=1 → salida: 32×32
    * MaxPool2: → 16×16
  * Flatten: 16×16×n_filtros
  * Dense → salida final: 1 (sigmoid)

Cálculo ejemplo (con 16 y 32 filtros):
* Input: 64×64×1
Conv1 (3x3, pad=1, stride=1, 16 filtros): 64×64×16
  * MaxPool1 (2x2, stride=2): 32×32×16
* Conv2 (3x3, pad=1, stride=1, 32 filtros): 32×32×32
  * MaxPool2 (2x2): 16×16×32
* Flatten: 8192
* Dense(128) + Dense(1, sigmoid)
