La fórmula general para el tamaño de salida es:
* Salida=⌊(𝑊−𝐾+2𝑃)/𝑆⌋+1

donde:
* 𝑊: dimensión de entrada (asumimos cuadrada)
* 𝐾: tamaño del kernel
* 𝑆: stride
* 𝑃: padding

Ejercicio:

Una imagen de entrada tiene dimensión 32×32

Se aplica una convolución con:
* Kernel => 𝐾=5
* Padding => 𝑃=0
* Stride => 𝑆=1

¿Cuál será la dimensión de salida?

Solución:

Salida=⌊(32−5+2×0)/1⌋+1=28

Salida: 28×28
