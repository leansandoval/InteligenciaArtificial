# Operación de convolución (sin padding y stride = 1)

Ejercicio 1:

Dada la siguiente imagen de entrada
𝐼 y el filtro 𝐾:

I=
[
  1 3 1

  2 1 0

  0 2 1
        ]
​,

K=
[
  0 1  

  1 0
      ]

Realizar la convolución válida (sin padding, stride = 1) entre la imagen y el kernel.

Solución 1:

* Tamaño de entrada: 3×3
* Tamaño del kernel: 2×2
* Salida: (3−2+1)×(3−2+1)=2×2

Aplicamos la operación de convolución:
* Región superior izquierda:
(1×0+2×1+3×1+1×0)=0+2+3+0=5

* Región superior derecha:
(2×0+0×1+1×1+2×0)=0+0+1+0=1

* Región inferior izquierda:
(3×0+1×1+1×1+0×0)=0+1+1+0=2

* Región inferior derecha:
(1×0+2×1+0×1+1×0)=0+2+0+0=2

Resultado final:

[ 5 1

  2 2 ]
