La fórmula general para el tamaño de salida es:

Salida=⌊(𝑊−𝐾+2𝑃)/𝑆⌋+1

donde:
* 𝑊: dimensión de entrada (asumimos cuadrada)
* 𝐾: tamaño del kernel
* 𝑆: stride
* 𝑃: padding

Ejercicio, ahora con:
* Imagen de entrada: 64×64
* Kernel: 3
* Padding: 1
* Stride: 1

Solución:

Salida=⌊(64−3+2×1)/1⌋+1=⌊63/1⌋+1=64

Salida: 64×64
