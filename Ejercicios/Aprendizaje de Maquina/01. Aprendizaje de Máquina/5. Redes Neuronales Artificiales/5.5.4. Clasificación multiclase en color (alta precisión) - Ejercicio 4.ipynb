Enunciado:

Queremos clasificar imágenes de 10 especies de flores.
Imágenes de color 128×128×3. Se busca buena precisión en PC con GPU.

Solución propuesta:
* Tipo de convolución: Conv2D, 3 canales
* Pooling: MaxPooling cada 2 bloques
* Activación: ReLU + softmax
* Capas ocultas: 3 bloques conv + Dense(256)
* Data Augmentation: sí (fuera del modelo)

Cálculo de salida:
* Input: 128×128×3  
* Conv1 (3×3, 32 filtros) → 128×128×32  
  * MaxPool1 → 64×64×32  
* Conv2 (3×3, 64 filtros) → 64×64×64  
  * MaxPool2 → 32×32×64  
* Conv3 (3×3, 128 filtros) → 32×32×128  
  * MaxPool3 → 16×16×128  
* Flatten → Dense(256) → Dense(10, softmax)
