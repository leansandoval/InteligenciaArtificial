A continuación te presento más ejercicios prácticos resueltos para:

* Determinar la cantidad de parámetros a entrenar en redes Perceptrón Multicapa (MLP) y Convolucionales (CNN).
* Comparar resultados numéricos y analizar el impacto en recursos computacionales y tiempo de procesamiento.

¿CÓMO CALCULAR PARÁMETROS?

* Capa densa (fully connected):
  * Parámetros=Entradas×Neuronas+Biases

* Capa convolucional (Conv2D):
  * Parámetros=Alto kernel×Ancho kernel×Canales entrada×Filtros+Filtros (biases)

# Calcular Parámetros Entrenables en un Perceptrón Multicapa (MLP)

Enunciado:

* Queremos clasificar imágenes de dígitos 28×28 en 10 clases usando una MLP.

Arquitectura propuesta:
* Entrada: 784 neuronas (28×28)
* Capa oculta 1: 128 neuronas, activación ReLU
* Capa oculta 2: 64 neuronas
* Salida: 10 neuronas (Softmax)

Solución paso a paso:
Entrada → Oculta 1:
(784 inputs×128 neuronas)+128 biases=100480

Oculta 1 → Oculta 2:
(128×64)+64=8256

Oculta 2 → Salida:
(64×10)+10=650

Total parámetros MLP:
100480+8256+650=109386 parámetros

# Calcular Parámetros Entrenables en una CNN

Enunciado:

Usamos la misma tarea (clasificar imágenes 28×28×1 en 10 clases), pero con una red convolucional:
* Conv2D (filtros=16, kernel=3×3, stride=1, padding='valid')
  * MaxPooling (2×2)
* Conv2D (filtros=32, kernel=3×3)
  * MaxPooling (2×2)
* Flatten → Dense(64) → Dense(10)

Cálculo de parámetros CNN:

* Conv1: (entrada 28×28×1 → salida 26×26×16)
(3×3×1)×16+16=144+16=160
* Conv2: (entrada 13×13×16 → salida 11×11×32)
(3×3×16)×32+32=4608+32=4640
* Flatten:
Después del segundo MaxPool → 5×5×32=800
* Dense(800 → 64):
800×64+64=51264
* Dense(64 → 10):
64×10+10=650

Total parámetros CNN: 160+4640+51264+650=56694 parámetros

# Comparativa Final

| Aspecto                                   | MLP                     | CNN                            |
| ----------------------------------------- | ----------------------- | ------------------------------ |
| Parámetros                                | 109,386                 | 56,694                         |
| Capacidad de capturar patrones espaciales | No                      | Sí (convolución)             |
| Eficiencia con imágenes grandes           | Baja                    |  Alta                         |
| Riesgo de overfitting                     | Alto (más parámetros)   |  Medio                       |
| Costo computacional                       | Más memoria y tiempo    | Menor por pesos compartidos |
| Recomendado para                          | Datos tabulares         | Imágenes, señales, texto       |
