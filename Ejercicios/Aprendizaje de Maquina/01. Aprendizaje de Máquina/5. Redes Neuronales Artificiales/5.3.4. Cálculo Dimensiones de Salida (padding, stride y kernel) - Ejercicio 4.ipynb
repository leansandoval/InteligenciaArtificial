La fórmula general para el tamaño de salida es:

Salida=⌊(𝑊−𝐾+2𝑃)/𝑆⌋+1

donde:
* 𝑊: dimensión de entrada (asumimos cuadrada)
* 𝐾: tamaño del kernel
* 𝑆: stride
* 𝑃: padding

Cálculo de salida con múltiples filtros y canales

Enunciado:
Una imagen de entrada de tamaño 32×32×3 (imagen RGB).
Se aplica una capa convolucional con:
* Kernel = 3×3
* Stride = 1
* Padding = 1
* Cantidad de filtros: 16

Solución:
* Usamos la fórmula para el tamaño de salida espacial (ancho y alto):
  * Salida=⌊(𝑊−𝐾+2𝑃)/𝑆⌋+1=⌊(32−3+2×1)/1⌋+1=32
  * Cada filtro opera sobre los 3 canales (uno para cada canal RGB), por lo que cada filtro tiene dimensiones 3×3×3=27 pesos + 1 bias.

Resultado final:
* Salida de dimensión: 32×32×16
