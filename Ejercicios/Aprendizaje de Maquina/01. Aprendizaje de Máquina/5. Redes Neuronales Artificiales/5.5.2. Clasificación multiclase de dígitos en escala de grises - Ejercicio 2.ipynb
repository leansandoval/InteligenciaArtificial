Enunciado:

Queremos clasificar imágenes de dígitos del 0 al 9 en escala de grises (como MNIST, 28×28 px).
Se desea una red pequeña pero precisa para dispositivos móviles.

Solución propuesta:
* Tipo de convolución: Conv2D, entrada 28×28×1
* Pooling: Sí, MaxPooling
* Activaciones: ReLU + softmax (10 clases)
* Capas ocultas: 2 bloques conv + FC

Cálculo de dimensiones:
* Input: 28×28×1
* Conv1: 3×3, stride=1, pad=1, 32 filtros → 28×28×32  
  * MaxPool1: 2×2 → 14×14×32  
* Conv2: 3×3, 64 filtros → 14×14×64  
  * MaxPool2 → 7×7×64  
* Flatten: 3136 → Dense(128) → Dense(10, softmax)
