Enunciado:

Queremos clasificar si una radiografía de tórax presenta o no neumonía.
Las imágenes son en color, tamaño 256×256. Se busca una arquitectura precisa, no limitada en recursos.

Solución propuesta:
* Tipo de convolución: Conv2D con 3 canales (RGB)
* Pooling: Sí, para reducir gradualmente
* Activaciones: ReLU + sigmoid (binaria)
* Capas ocultas: 4 bloques conv + 2 densas
* Regularización: Dropout

Arquitectura resumida:
* Input: 256×256×3
* Conv1 (3×3, 32 filtros) → 256×256×32  
  * MaxPool → 128×128×32  
* Conv2 (64 filtros) → 128×128×64  
  * MaxPool → 64×64×64  
* Conv3 (128 filtros) → 64×64×128  
  * MaxPool → 32×32×128  
* Conv4 (256 filtros) → 32×32×256  
  * MaxPool → 16×16×256  
* Flatten → Dense(512) + Dropout → Dense(1, sigmoid)
