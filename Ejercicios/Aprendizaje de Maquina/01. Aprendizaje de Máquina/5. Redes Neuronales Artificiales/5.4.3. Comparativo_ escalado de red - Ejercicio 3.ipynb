Enunciado:
Se desea clasificar imágenes de frutas.

Tenés imágenes:
* En escala de grises: 32×32 px
* En color: 128×128 px

¿Cuál arquitectura sugerís para cada caso?

| Característica      | Escala de grises   | Color          |
| ------------------- | ------------------ | -------------- |
| Tamaño entrada      | 32×32×1            | 128×128×3      |
| Filtros             | 8–16–32            | 32–64–128      |
| Capas ocultas       | 2 conv + 1 FC      | 3 conv + 2 FC  |
| Pooling             | Sí, tras cada conv | Sí, cada 2     |
| Activaciones        | ReLU + sigmoid     | ReLU + softmax |
| Coste computacional | Bajo               | Medio-Alto     |
