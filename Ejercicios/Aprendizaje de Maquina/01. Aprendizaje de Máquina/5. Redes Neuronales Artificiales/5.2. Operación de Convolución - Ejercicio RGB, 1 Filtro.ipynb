# Convolución sobre imagen RGB con 1 filtro

Enunciado:

Dada una imagen de entrada de tamaño 4×4 con 3 canales (RGB) y

un filtro convolucional de tamaño 2×2×3 (2x2 en espacio, 3 canales),

sin padding y con stride = 1.

Supongamos que el filtro tiene pesos:

* Filtro (2x2x3):

image.png

* La imagen (canales separados):

image.png

Solución paso a paso:
* Primero, calculamos la salida de la convolución en una sola posición:
* Ventana inicial (posición 0,0):

Canal R:
(1×1)+(2×0)+(2×0)+(3×−1)=1+0+0−3=−2

Canal G:
(0×0)+(1×1)+(1×1)+(2×0)=0+1+1+0=2

Canal B:
(1×1)+(1×−1)+(0×0)+(2×0)=1−1+0+0=0

Total suma: −2+2+0=0

Repetís esto en cada posición para obtener una matriz de salida 3×3
