La fórmula general para el tamaño de salida es:

Salida=⌊(𝑊−𝐾+2𝑃)/𝑆⌋+1

donde:
* 𝑊: dimensión de entrada (asumimos cuadrada)
* 𝐾: tamaño del kernel
* 𝑆: stride
* 𝑃: padding

Ejercicio:
* Imagen de entrada: 128×128
* Kernel: 7
* Padding: 0
* Stride: 2

Solución:

Salida=⌊(128−7+2×0)/2⌋+1=⌊121/2⌋+1=60+1=61

Aclaración: En este caso la división 121/2 nos da como resultado 60,5
=> Debemo truncar valor entero => 60

Salida: 61×61
