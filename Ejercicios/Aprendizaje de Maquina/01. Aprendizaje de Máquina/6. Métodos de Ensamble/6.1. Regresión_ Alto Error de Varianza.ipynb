Problema: Predecir el precio de una casa a partir de características estructurales y ubicación (datos tabulares).

Primer modelo: Árbol de decisión profundo.

Resultados:

* Entrenamiento: RMSE = 12.000
* Validación: RMSE = 28.000

Diagnóstico: Overfitting → Alta varianza, bajo sesgo.

* Objetivo: Disminuir la varianza manteniendo el sesgo bajo.
* Técnica sugerida: Bagging (Random Forest)

Justificación:
* El árbol de decisión tiene alta varianza por su sensibilidad a los datos de entrenamiento.
* El bagging (como Random Forest) entrena múltiples árboles en subconjuntos bootstrap para promediar sus predicciones, reduciendo la varianza.

¿CUÁNDO USAR CADA MÉTODO DE ENSAMBLE?

| Situación del modelo base    | Método de ensamble sugerido | Justificación                                |
| ---------------------------- | --------------------------- | -------------------------------------------- |
| Alta varianza (overfitting)  | Bagging (Random Forest)     | Reduce varianza promediando                  |
| Alto sesgo (underfitting)    | Boosting (AdaBoost/XGBoost) | Reduce sesgo corrigiendo errores             |
| Error distribuido (moderado) | Stacking                    | Aprovecha fortalezas combinadas              |
| Ruido alto en datos          | Ninguno (mejorar calidad)   | Error irreducible no se corrige con ensamble |
