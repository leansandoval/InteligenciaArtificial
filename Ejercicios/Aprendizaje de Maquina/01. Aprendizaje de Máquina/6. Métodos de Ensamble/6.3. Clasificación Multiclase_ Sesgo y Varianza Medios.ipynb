Problema: Clasificación de imágenes de ropa (Fashion-MNIST).

Primer modelo: MLP básico (1 capa oculta).

Resultados:

* Accuracy entrenamiento: 82%
* Accuracy validación: 78%

Diagnóstico: Error total razonable, pero con margen de mejora en sesgo y varianza.
* Objetivo: Mejorar performance aprovechando modelos heterogéneos.
* Técnica sugerida: Stacking (Ensemble heterogéneo)

Justificación:
* Stacking permite combinar distintos tipos de modelos (CNN, MLP, KNN, árboles, etc.) y usar un metamodelo que aprenda a ponderar sus salidas.
* Aporta robustez cuando diferentes modelos capturan diferentes patrones de los datos.
* Útil cuando no hay un claro problema de sesgo o varianza predominante, pero sí una oportunidad de mejora global.

¿CUÁNDO USAR CADA MÉTODO DE ENSAMBLE?

| Situación del modelo base    | Método de ensamble sugerido | Justificación                                |
| ---------------------------- | --------------------------- | -------------------------------------------- |
| Alta varianza (overfitting)  | Bagging (Random Forest)     | Reduce varianza promediando                  |
| Alto sesgo (underfitting)    | Boosting (AdaBoost/XGBoost) | Reduce sesgo corrigiendo errores             |
| Error distribuido (moderado) | Stacking                    | Aprovecha fortalezas combinadas              |
| Ruido alto en datos          | Ninguno (mejorar calidad)   | Error irreducible no se corrige con ensamble |
