Problema: Clasificar si un paciente tiene una enfermedad rara (5% positivos).

Primer modelo: SVM con kernel lineal.

Resultados:

* Precision (positivos): 30%
* Recall (positivos): 40%

Diagnóstico: El modelo subajusta los positivos (alto sesgo en clase minoritaria).
* Objetivo: Mejorar recall y precisión de la clase minoritaria.
* Técnica sugerida: Boosting (XGBoost con weight_balance)

Justificación:
* Boosting maneja bien los errores secuenciales.
* Se puede configurar para dar más peso a los errores en la clase minoritaria, corrigiendo el sesgo.
* Mejora la detección de la clase positiva sin arriesgar sobreajuste.

¿CUÁNDO USAR CADA MÉTODO DE ENSAMBLE?

| Situación del modelo base    | Método de ensamble sugerido | Justificación                                |
| ---------------------------- | --------------------------- | -------------------------------------------- |
| Alta varianza (overfitting)  | Bagging (Random Forest)     | Reduce varianza promediando                  |
| Alto sesgo (underfitting)    | Boosting (AdaBoost/XGBoost) | Reduce sesgo corrigiendo errores             |
| Error distribuido (moderado) | Stacking                    | Aprovecha fortalezas combinadas              |
| Ruido alto en datos          | Ninguno (mejorar calidad)   | Error irreducible no se corrige con ensamble |
