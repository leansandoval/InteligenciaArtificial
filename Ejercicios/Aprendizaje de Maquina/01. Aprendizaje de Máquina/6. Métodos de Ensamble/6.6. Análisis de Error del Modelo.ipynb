# Interpretación y elección del método de ensamble en cada caso

1. Regresión
* Modelo base tiene varianza alta (barra naranja alta).
* Bagging reduce considerablemente la varianza.
* Boosting mejora el sesgo, pero mantiene varianza moderada.
* Stacking ofrece un balance intermedio.

Elección sugerida: Bagging (Random Forest)
Justificación: El modelo base sufre de overfitting. Bagging reduce varianza sin aumentar el sesgo.

2. Clasificación Binaria
* Modelo base muestra sesgo alto (barra azul muy alta).
* Boosting reduce notablemente el sesgo.
* Bagging apenas mejora.
* Stacking es competitivo, pero no tan eficaz en este caso como Boosting.

Elección sugerida: Boosting (AdaBoost o XGBoost)
Justificación: El underfitting del modelo base es evidente. Boosting ataca directamente el problema del sesgo.

3. Clasificación Multiclase
* Modelo base con errores equilibrados, pero mejorables.
* Boosting y Stacking logran mejoras tanto en sesgo como varianza.
* Bagging es menos efectivo aquí.

Elección sugerida: Stacking (Ensamble heterogéneo)
Justificación: El error es moderado en ambas fuentes. Stacking combina fortalezas de modelos diversos, ideal para problemas multiclase complejos.

# Consideraciones para elegir una técnica de ensamble

| Criterio                      | Bagging             | Boosting          | Stacking                                  |
| ----------------------------- | ------------------- | ----------------- | ----------------------------------------- |
| Alta varianza (overfitting)   | Muy efectivo      | Puede aumentar | Moderado                               |
| Alto sesgo (underfitting)     | No mejora sesgo   | Muy efectivo    | Moderado                               |
| Error distribuido equilibrado | Mejora parcial   | Mejora parcial | Recomendado                             |
| Tarea multiclase compleja     | Funciona regular | Funciona bien   | Óptimo si se combinan modelos distintos |
| Interpretabilidad             | Alta (RF)         | Media-baja     | Baja                                   |
