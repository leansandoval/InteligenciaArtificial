Problema: Predecir emisiones de CO₂ según múltiples sensores ambientales.

Primer modelo: Regresión lineal múltiple.

Resultados:

* Entrenamiento: MSE = 3.5
* Validación: MSE = 3.8

Diagnóstico: Bajo sesgo y varianza, pero con error irreducible alto (ruido en datos).
* Objetivo: No necesariamente ensamble, sino reconocer el límite del modelo.
* Técnica sugerida: No se justifica ensamble. En su lugar:
* Mejorar calidad de datos (limpieza, eliminación de outliers).
* Aplicar técnicas de regularización o análisis de errores residuales.

¿CUÁNDO USAR CADA MÉTODO DE ENSAMBLE?

| Situación del modelo base    | Método de ensamble sugerido | Justificación                                |
| ---------------------------- | --------------------------- | -------------------------------------------- |
| Alta varianza (overfitting)  | Bagging (Random Forest)     | Reduce varianza promediando                  |
| Alto sesgo (underfitting)    | Boosting (AdaBoost/XGBoost) | Reduce sesgo corrigiendo errores             |
| Error distribuido (moderado) | Stacking                    | Aprovecha fortalezas combinadas              |
| Ruido alto en datos          | Ninguno (mejorar calidad)   | Error irreducible no se corrige con ensamble |
