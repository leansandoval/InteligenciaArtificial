Problema: Clasificar si un correo es spam o no, usando Naive Bayes.

Primer modelo: Naive Bayes.

Resultados:

* Accuracy entrenamiento: 71%
* Accuracy validación: 70%

Diagnóstico: Underfitting → Alto sesgo, baja varianza.

* Objetivo: Disminuir el sesgo sin aumentar mucho la varianza.
* Técnica sugerida: Boosting (AdaBoost o Gradient Boosting)

Justificación:
* Boosting combina débilmente clasificadores (como árboles pequeños) de forma secuencial para mejorar el desempeño corrigiendo errores anteriores, ideal en modelos que subajustan (alto sesgo).
* Puede reducir el sesgo sin sobreajustar demasiado si se controla el número de etapas y profundidad de los árboles.

¿CUÁNDO USAR CADA MÉTODO DE ENSAMBLE?

| Situación del modelo base    | Método de ensamble sugerido | Justificación                                |
| ---------------------------- | --------------------------- | -------------------------------------------- |
| Alta varianza (overfitting)  | Bagging (Random Forest)     | Reduce varianza promediando                  |
| Alto sesgo (underfitting)    | Boosting (AdaBoost/XGBoost) | Reduce sesgo corrigiendo errores             |
| Error distribuido (moderado) | Stacking                    | Aprovecha fortalezas combinadas              |
| Ruido alto en datos          | Ninguno (mejorar calidad)   | Error irreducible no se corrige con ensamble |
