```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.naive_bayes import GaussianNB
```

```python
# Assuming you have your data loaded into variables named X and y
# Replace these lines with your actual data loading and preparation
# For example, if loading from a CSV using pandas:
# import pandas as pd
# df = pd.read_csv('your_data.csv')
# X = df[['feature1', 'feature2']].values # Replace with your feature column names
# y = df['target'].values # Replace with your target column name

# For demonstration purposes, let's create some dummy data
from sklearn.datasets import make_classification
X, y = make_classification(n_samples=100, n_features=2, n_informative=2, n_redundant=0, random_state=42)
```

```python
# Modelo Naive Bayes
model = GaussianNB()
model.fit(X, y)
```

```python
# Gráfico de dispersión con frontera de decisión
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

# Crear una malla para la frontera de decisión
xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 100), np.linspace(ylim[0], ylim[1], 100))
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Dibujar la frontera de decisión
plt.contour(xx, yy, Z, levels=[0], linewidths=2, colors='black')
plt.title('Naive Bayes')
plt.show()
```
