```python
import numpy as np # Import numpy
import matplotlib.pyplot as plt # Import matplotlib
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_classification
```

```python
# Generación de datos para clasificación binaria
X, y = make_classification(n_samples=100, n_features=2, n_classes=2, n_clusters_per_class=1, n_informative=2, n_redundant=0, n_repeated=0)
```

```python
# Modelo Árbol de Decisión
model = DecisionTreeClassifier(max_depth=3)
model.fit(X, y)
```

```python
# Gráfico de dispersión con frontera de decisión
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

# Crear una malla para la frontera de decisión
xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 100), np.linspace(ylim[0], ylim[1], 100))
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Dibujar la frontera de decisión
plt.contour(xx, yy, Z, levels=[0], linewidths=2, colors='black')
plt.title('Árbol de Decisión')
plt.show()
```
