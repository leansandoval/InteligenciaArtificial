```python
import numpy as np
from sklearn.svm import SVC
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
```

```python
# Generación de datos para clasificación binaria
# Set n_informative to be <= n_features, and n_redundant and n_repeated to 0
X, y = make_classification(n_samples=100, n_features=2, n_informative=2, n_redundant=0, n_repeated=0, n_classes=2, n_clusters_per_class=1)
```

```python
# Modelo SVM
model = SVC(kernel='linear')
model.fit(X, y)
```

```python
# Gráfico de dispersión con frontera de decisión
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

# Crear una malla para la frontera de decisión
xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 100), np.linspace(ylim[0], ylim[1], 100))
Z = model.decision_function(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Dibujar la frontera de decisión
plt.contour(xx, yy, Z, levels=[0], linewidths=2, colors='black')
plt.title('SVM con Kernel Lineal')
plt.show()
```
