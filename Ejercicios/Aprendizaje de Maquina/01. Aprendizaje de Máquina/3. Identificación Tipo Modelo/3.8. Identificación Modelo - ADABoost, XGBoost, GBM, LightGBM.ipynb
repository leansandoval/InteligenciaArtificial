```python
import numpy as np # Importar numpy
import matplotlib.pyplot as plt # Importar matplotlib
from sklearn.datasets import make_classification # Importar make_classification para datos de ejemplo
from sklearn.ensemble import GradientBoostingClassifier
```

```python
# Generar datos de ejemplo (reemplazar con tus datos reales)
X, y = make_classification(n_samples=100, n_features=2, n_informative=2, n_redundant=0, random_state=42)
```

```python
# Modelo GBM
model = GradientBoostingClassifier(n_estimators=100)
model.fit(X, y)
```

```python
# Gráfico de dispersión con frontera de decisión
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

# Crear una malla para la frontera de decisión
xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 100), np.linspace(ylim[0], ylim[1], 100))
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Dibujar la frontera de decisión
plt.contour(xx, yy, Z, levels=[0], linewidths=2, colors='black')
plt.title('GBM')
plt.show()
```
