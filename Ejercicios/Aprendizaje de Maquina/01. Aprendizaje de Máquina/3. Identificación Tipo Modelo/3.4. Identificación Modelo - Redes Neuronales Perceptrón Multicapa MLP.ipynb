```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPClassifier
```

```python
# Define X and y (Replace with your actual data loading/generation)
# Example using make_classification for dummy data
from sklearn.datasets import make_classification
X, y = make_classification(n_samples=100, n_features=2, n_informative=2, n_redundant=0, random_state=42)
```

```python
# Modelo de Red Neuronal (MLP)
model = MLPClassifier(hidden_layer_sizes=(5,), max_iter=1000)
model.fit(X, y)
```

```python
# Gráfico de dispersión con frontera de decisión
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

# Crear una malla para la frontera de decisión
xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 100), np.linspace(ylim[0], ylim[1], 100))
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Dibujar la frontera de decisión
plt.contour(xx, yy, Z, levels=[0], linewidths=2, colors='black')
plt.title('Red Neuronal (MLP)')
plt.show()
```
