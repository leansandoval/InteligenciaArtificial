```python
import numpy as np # Import numpy
import matplotlib.pyplot as plt # Import matplotlib
from sklearn.linear_model import LogisticRegression
```

```python
# Generación de datos para clasificación binaria (ajustado para un solo feature y visualización)
np.random.seed(42) # for reproducibility
X = np.random.rand(100, 1) * 6 - 3 # Generate 100 data points between -3 and 3
y = (X.squeeze() + np.random.randn(100) * 0.5 > 0).astype(int) # Simple linear relationship + noise, convert to binary
```

```python
# Modelo de regresión logística
model = LogisticRegression()
model.fit(X, y)
```

```python
# Predicciones
X_test = np.linspace(-3, 3, 100).reshape(-1, 1)
y_pred = model.predict_proba(X_test)[:, 1]
```

```python
# Gráfico
plt.scatter(X, y, color='blue', label='Datos')
plt.plot(X_test, y_pred, color='red', label='Regresión Logística')
plt.title('Regresión Logística')
plt.xlabel('X')
plt.ylabel('Probabilidad')
plt.legend()
plt.show()
```
