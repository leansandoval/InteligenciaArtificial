```python
import numpy as np
import gym
```

```python
# Crear el entorno FrozenLake 4x4
env = gym.make("FrozenLake-v1", is_slippery=False)
```

```python
# Inicializar la tabla Q con ceros
Q_table = np.zeros((env.observation_space.n, env.action_space.n))
```

```python
# Parámetros de Q-Learning
alpha = 0.1      # Tasa de aprendizaje
gamma = 0.99     # Factor de descuento
epsilon = 1.0    # Exploración inicial
epsilon_decay = 0.995
epsilon_min = 0.01
episodes = 2000
```

```python
# Entrenamiento con Q-Learning
for episode in range(episodes):
    # The env.reset() function now returns a tuple: (observation, info)
    # We only need the observation, which is the first element
    state = env.reset()
    # Check if state is an int and if so, do not try to access it as a list
    if isinstance(state, int):
        pass # Or assign state directly, depending on your logic
    else:
        state = state[0] # If it's not an int, it should be a tuple/list

    done = False

    while not done:
        # Seleccionar acción con política epsilon-greedy
        if np.random.rand() < epsilon:
            action = env.action_space.sample()  # Exploración
        else:
            action = np.argmax(Q_table[state, :])  # Explotación

        # Ejecutar acción en el entorno
        next_state, reward, done, _, _ = env.step(action)

        # Actualizar Q-Table con la ecuación de Bellman
        Q_table[state, action] += alpha * (reward + gamma * np.max(Q_table[next_state, :]) - Q_table[state, action])

        state = next_state

    # Reducir epsilon (menos exploración con el tiempo)
    epsilon = max(epsilon * epsilon_decay, epsilon_min)
```

```python
print("\nTabla Q aprendida:\n", Q_table)
```
