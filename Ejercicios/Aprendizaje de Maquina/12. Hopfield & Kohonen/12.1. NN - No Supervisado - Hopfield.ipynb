```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class HopfieldNetwork:
    def __init__(self, n_neurons):
        self.n_neurons = n_neurons
        self.weights = np.zeros((n_neurons, n_neurons))

    def train(self, patterns):
        """Entrenar la red con los patrones almacenados usando la regla de Hebb."""
        for pattern in patterns:
            self.weights += np.outer(pattern, pattern)
        np.fill_diagonal(self.weights, 0)  # Sin autoconexiones

    def recall(self, pattern, steps=5):
        """Recuperar un patrón a partir de una entrada con ruido."""
        recalled_pattern = pattern.copy()
        for _ in range(steps):
            for i in range(self.n_neurons):
                recalled_pattern[i] = np.sign(np.dot(self.weights[i], recalled_pattern))
        return recalled_pattern
```

```python
# Definir patrones de entrenamiento (memoria)
pattern1 = np.array([1, -1, 1, -1, 1, -1])
pattern2 = np.array([-1, 1, -1, 1, -1, 1])
patterns = np.array([pattern1, pattern2])
```

```python
# Crear y entrenar la red de Hopfield
hopfield_net = HopfieldNetwork(n_neurons=6)
hopfield_net.train(patterns)
```

```python
# Introducir ruido en un patrón
noisy_pattern = np.array([1, -1, 1, -1, -1, -1])  # Se modificaron 2 valores
recovered_pattern = hopfield_net.recall(noisy_pattern)
```

```python
print("Patrón original:", pattern1)
print("Patrón ruidoso:", noisy_pattern)
print("Patrón recuperado:", recovered_pattern)
```

Explicación del código:

* Se definen dos patrones binarios de memoria.
* Se entrena una Red de Hopfield con la regla de Hebb.
* Se introduce ruido en un patrón y la red intenta recuperarlo.
* Se imprime la comparación entre el patrón original, el distorsionado y el recuperado.
