Keras

Importación de librerías.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.datasets import mnist
```

Carga de los datos y división en subconjuntos de entrenamiento y prueba.

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

Normalizamos los datos.

```python
x_train, x_test = x_train / 255.0, x_test / 255.0
```

Conformamos la arquitectura de nuestra red neuronal perceptrón simple.

```python
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(10, activation='softmax')  # Perceptrón simple
])
```

Compilamos modelo.

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

Entrenamos modelo.

```python
model.fit(x_train, y_train, epochs=10, validation_data=(x_test, y_test))
```

PyTorch

Importamos librerías.

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
```

Cargamos los subconjuntos de entrenamiento y prueba con datos correspondientes.

```python
transform = transforms.ToTensor()
train_loader = torch.utils.data.DataLoader(datasets.MNIST('.', train=True, download=True, transform=transform), batch_size=64)
test_loader = torch.utils.data.DataLoader(datasets.MNIST('.', train=False, transform=transform), batch_size=64)
```

Confeccionamos nuestro modelo.

```python
class Perceptron(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(28*28, 10)

    def forward(self, x):
        x = x.view(-1, 28*28)
        return self.fc(x)
```

Instanciamos y configuramos nuestro modelo con unos determinados hiperparámetros como la función de pérdida (CrossEntropyLoss) y técnica de optimización (Adam).

```python
model = Perceptron()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

Entrenamos modelo.

```python
for epoch in range(5):
    for x, y in train_loader:
        optimizer.zero_grad()
        output = model(x)
        loss = criterion(output, y)
        loss.backward()
        optimizer.step()
```
