```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules, fpgrowth
```

```python
# Crear dataset de transacciones
data = pd.DataFrame([
    ['Leche', 'Pan', 'Mantequilla'],
    ['Pan', 'Mantequilla'],
    ['Leche', 'Pan'],
    ['Leche', 'Mantequilla'],
    ['Pan', 'Leche', 'Mantequilla', 'Queso']
])
```

```python
# Convertir transacciones a formato One-Hot Encoding
from mlxtend.preprocessing import TransactionEncoder
```

```python
# Reemplazar valores None con cadenas vacías
data = data.fillna('')
```

```python
# Convertir transacciones a formato One-Hot Encoding
te = TransactionEncoder()
data_encoded = te.fit(data.values).transform(data.values)
df = pd.DataFrame(data_encoded, columns=te.columns_)
```

```python
# Aplicar FP-Growth para encontrar patrones frecuentes
frequent_itemsets = fpgrowth(df, min_support=0.5, use_colnames=True)
```

```python
# Generar reglas de asociación
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)
```

```python
print("\n\tPatrones Frecuentes:")
print(frequent_itemsets)
```

```python
print("\n\tReglas de Asociación:")
print(rules)
```
