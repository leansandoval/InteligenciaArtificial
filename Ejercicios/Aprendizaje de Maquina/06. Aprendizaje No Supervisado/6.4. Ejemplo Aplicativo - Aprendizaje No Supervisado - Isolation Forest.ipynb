```python
import numpy as np
from sklearn.ensemble import IsolationForest
import matplotlib.pyplot as plt
```

```python
# Generar datos de ejemplo: datos normales y anomalías
X_normal = np.random.normal(loc=0, scale=1, size=(100, 2))  # Datos normales (distribución normal)
X_outliers = np.random.uniform(low=-4, high=4, size=(10, 2))  # Outliers aleatorios
X = np.vstack([X_normal, X_outliers])  # Combina los datos
```

```python
# Crear el modelo Isolation Forest
model = IsolationForest(contamination=0.1)  # 10% de los datos son outliers
```

```python
# Entrenar el modelo
model.fit(X)
```

```python
# Predecir anomalías
y_pred = model.predict(X)
```

```python
# Visualizar los resultados
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='coolwarm')
plt.title("Detección de Anomalías con Isolation Forest")
plt.xlabel("Característica 1")
plt.ylabel("Característica 2")
plt.show()
```

Explicación del código:

* Se generan datos normales y outliers aleatorios.
* Se entrena un modelo Isolation Forest con un parámetro de contaminación (proporción de outliers esperados en los datos).
* Se predicen los outliers usando model.predict(), donde -1 indica outliers y 1 indica datos normales.
* Se visualizan los puntos con un color que indica si son normales o outliers.
